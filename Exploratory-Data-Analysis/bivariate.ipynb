```python
# import packages
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
from scipy.stats.contingency import association
import seaborn as sns
```

```python
# load the churn data as a dataframe
df = pd.read_csv('churn_clean.csv')
```

```python
# describe MonthlyCharge
df.MonthlyCharge.describe()
```

```python
# describe Bandwidth_GB_Year
df.Bandwidth_GB_Year.describe()
```

```python
# calculate the correlation coefficient between MonthlyCharge and Bandwidth_GB_Year
correlation = stats.pearsonr(df.MonthlyCharge, df.Bandwidth_GB_Year)
print(correlation)
```

```python
# scatterplot of MonthlyCharge and Bandwidth_GB_Year; two distinct groups, could be based on factors such as price locks
sns.scatterplot(x = 'Bandwidth_GB_Year', y = 'MonthlyCharge', data = df)
plt.show()
```

```python
# describe Techie
df.Techie.describe()
```

```python
# describe Churn
df.Churn.describe()
```

```python
# create cross tabulation for Techie and Churn
cross_tab = pd.crosstab(df.Techie, df.Churn)
print(cross_tab)
```

```python
# plot Techie and Churn; distribution shows when someone is not a techie they also haven't churned
sns.countplot(x = 'Techie', hue = 'Churn', data = df)
plt.show()
```

```python
# calculate association using cramer's v
association(cross_tab, method = 'cramer')
```
