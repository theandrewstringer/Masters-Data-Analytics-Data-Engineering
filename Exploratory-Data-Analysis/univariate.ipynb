```python
# import packages
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
import seaborn as sns
```

```python
# load the churn data as a dataframe
df = pd.read_csv('churn_clean.csv')
```

```python
# describe Outage_sec_perweek
df.Outage_sec_perweek.describe()
```

```python
# calculate skewness
outage_sec_perweek_skewness = stats.skew(df.Outage_sec_perweek)
print(outage_sec_perweek_skewness)
```

```python
# histogram of Outage_sec_perweek; skewness indicates normal distribution
sns.histplot(df.Outage_sec_perweek)
plt.show()
```

```python
# describe Bandwidth_GB_Year
df.Bandwidth_GB_Year.describe()
```

```python
# calculate skewness
bandwidth_gb_year_skewness = stats.skew(df.Bandwidth_GB_Year)
print(bandwidth_gb_year_skewness)
```

```python
# histogram of Bandwidth_GB_Year; skewness indicates normal distribution
sns.histplot(df.Bandwidth_GB_Year)
plt.show()
```

```python
# create count values for the categorical variables for graphing purposes
area_counts = df.Area.value_counts()
contract_counts = df.Contract.value_counts()
```

```python
# describe Area
df.Area.describe()
```

```python
# frequency table for Area
print(area_counts)
```

```python
# bar plot for Area
sns.barplot(x = area_counts.index, y = area_counts.values)
plt.show()
```

```python
# describe Contract
df.Contract.describe()
```

```python
# frequency table for Contract
print(contract_counts)
```

```python
# bar plot for Contracts
sns.barplot(x = contract_counts.index, y = contract_counts.values)
plt.show()
```
