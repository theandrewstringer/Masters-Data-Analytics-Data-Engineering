```python
# import packages
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
import seaborn as sns
```

```python
# import the cleaned churn data to a dataframe
df = pd.read_csv('churn_clean.csv')
```

```python
# describe area
df.Area.describe()
```

```python
# describe outage_sec_perweek
df.Outage_sec_perweek.describe()
```

```python
# view boxplots of the area and outage_sec_perweek columns
sns.boxplot(x='Area', y='Outage_sec_perweek', data=df)
plt.show()
```

```python
# create the arrays for the anova test
rural_outages = df[df.Area == 'Rural'].Outage_sec_perweek
urban_outages = df[df.Area == 'Urban'].Outage_sec_perweek
suburban_outages = df[df.Area == 'Suburban'].Outage_sec_perweek
```

```python
# run the anova test
anova = stats.f_oneway(rural_outages, urban_outages, suburban_outages)
print(anova)
```
